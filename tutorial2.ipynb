```python
import numpy as np
```

```python
#question 1: create a numpy array of size 2 x 2
arr1 = np.array ([[1,2],[3,4]])
print("array (2,2) : \n", arr1)
```

```python
#question 2: create a random numpy array of size 3 x 4
arr2 = np.random.rand(3,4)
print("random numpy array (3,4): \n", arr2)
```

```python
#question 3: create a numpy array of ones and a numpy array of zeroes

arrones = np.ones((2, 3))
arrzeros = np.zeros((2, 3))

print("Array of ones: \n", arrones)
print("\nArray of zeros: \n", arrzeros)
```

```python
#question 4: create a random numpy array of size 6x3 and transpose it

arr4 = np.random.rand(6,3)
arrtrans = arr4.T

print("Original array (6,3): \n",arr4)
print("\nTransposed array (3,6): \n", arrtrans)
```

```python
#question 5: create two random numpy array of size n x p and p x m and then compute the dot product

n = 3
m = 5
p = 2

arrA = np.random.rand(n,p)
arrB = np.random.rand(p,m)
dotproduct = np.dot(arrA, arrB)

print("Array A: \n", arrA)
print("\nArray B: \n", arrB)
print("\nDot product of array A and array B: \n", dotproduct)
print("\nShape of array A: ", arrA.shape)
print("Shape of array B: ", arrB.shape)
print("Shape of dot product of array A and array B: ", dotproduct.shape)
```
